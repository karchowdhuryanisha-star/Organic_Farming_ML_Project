```python
X = df.drop(columns=[
"Sample_ID",
"Suitability_Score",
"Organic_Farming_Suitability"
])

y = df["Organic_Farming_Suitability"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(include=["number"]).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train)
print("Testing samples:", len(X_test))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    
)

print("Preprocessor created successfully!")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

print("Preprocessor created successfully!")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, random_state=42))
])

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
y_pred_lr = logistic_model.predict(X_test)

y_prob_lr = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy_lr = accuracy_score(y_test, y_pred_lr)

precision_lr = precision_score(
    y_test, y_pred_lr, pos_label="Suitable"

)
recall_lr = recall_score(
    y_test, y_pred_lr, pos_label="Suitable"
)

f1_lr = f1_score(
    y_test, y_pred_lr, pos_label="Suitable"
)

roc_auc_lr = roc_auc_score(
    (y_test == "Suitable").astype(int),
    y_prob_lr
)

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(accuracy_lr, 4))
print("Precision:", round(precision_lr, 4))
print("Recall   :", round(recall_lr, 4))
print("F1-score :", round(f1_lr, 4))
print("ROC-AUC  :", round(roc_auc_lr, 4))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_lr,
    labels=["Not Suitable", "Suitable"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_predictions(
    (y_test == "Suitable").astype(int),
    y_prob_lr
)

plt.title("ROC Curve - Logistic Regression")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

decision_tree_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=5,
        min_samples_split=10,
        min_samples_leaf=5,
        random_state=42
    ))
])

decision_tree_model.fit(X_train, y_train)

print("Decision Tree trained successfully!")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

random_forest_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        max_depth=8,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    ))
])

random_forest_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

knn_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", KNeighborsClassifier())
])

param_grid = {
    "classifier__n_neighbors": [3, 5, 7, 9, 11, 13, 15],
    "classifier__weights": ["uniform", "distance"],
    "classifier__p": [1, 2]
}

grid_knn = GridSearchCV(
    knn_pipeline,
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_knn.fit(X_train, y_train)

knn_model = grid_knn.best_estimator_

print("Best KNN parameters:", grid_knn.best_params_)
print("Best CV accuracy:", round(grid_knn.best_score_, 4))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

y_test_binary = (y_test == "Suitable").astype(int)

models_prob = {
    "Logistic Regression": y_prob_lr,
    "Decision Tree": y_prob_dt,
    "Random Forest": y_prob_rf,
    "KNN": y_prob_knn
}

plt.figure(figsize=(9, 6))

for model_name, probabilities in models_prob.items():
    fpr, tpr, _ = roc_curve(y_test_binary, probabilities)
    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{model_name} (AUC = {roc_auc:.3f})"
    )

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison of Classification Models")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Best Model Selection and Interpretation

Four machine learning classification algorithms—Logistic Regression, Decision Tree, Random Forest, and K-Nearest Neighbors (KNN)—were evaluated for predicting organic farming suitability.

Among the evaluated models, Logistic Regression achieved the best overall performance, with an accuracy of 80.50%, precision of 78.75%, recall of 74.12%, F1-score of 76.36%, and ROC-AUC of 0.8891. Random Forest also showed competitive performance, particularly with a ROC-AUC of 0.8858, but its overall test-set performance was slightly lower than Logistic Regression.

Decision Tree and KNN produced comparatively lower results. Therefore, Logistic Regression was selected as the best model for this dataset because it provided the strongest overall balance across accuracy, precision, recall, F1-score, and ROC-AUC.

These results demonstrate that model selection should not be based only on accuracy. Multiple evaluation metrics were considered to provide a more complete assessment of classification performance.


### Dataset Limitation

The dataset used in this project is a **source-calibrated synthetic dataset containing 1,000 instances**. It does not represent 1,000 authentic field or laboratory observations. The feature ranges were designed for educational purposes using realistic agricultural and soil-health parameters.

Furthermore, the `Organic_Farming_Suitability` target is derived from an instructional suitability rubric and should not be interpreted as an official organic farming certification or legal standard.

Therefore, the results of this project are suitable for demonstrating preprocessing, exploratory data analysis, classification, model comparison, evaluation, and explainability techniques. For real-world deployment, the models should be trained and validated using authenticated field and laboratory data with expert or agronomist-validated suitability labels.


### Conclusion

This project developed a machine learning-based approach for predicting whether agricultural conditions are suitable for organic farming. The workflow included data understanding, preprocessing, exploratory data analysis (EDA), train-test splitting, model training, evaluation, model comparison, and feature importance analysis.

Four classification algorithms were evaluated: Logistic Regression, Decision Tree, Random Forest, and K-Nearest Neighbors (KNN). Logistic Regression achieved the best overall performance with an accuracy of 80.50%, precision of 78.75%, recall of 74.12%, F1-score of 76.36%, and ROC-AUC of 0.8891. Random Forest also performed competitively, achieving an accuracy of 78.50% and ROC-AUC of 0.8858.

Based on the overall evaluation metrics, Logistic Regression was selected as the best-performing model for this dataset. The results also demonstrate the importance of considering precision, recall, F1-score, and ROC-AUC rather than relying only on accuracy.

Feature importance analysis using Random Forest provided additional insight into the variables influencing the predictions. However, because the dataset is source-calibrated synthetic data and the suitability labels are derived from an instructional rubric, the resulting model should be considered an educational demonstration rather than a system ready for real-world agricultural decision-making.

Future work could use authenticated field and laboratory observations with expert-validated suitability labels to develop and validate a more reliable model for practical organic farming suitability assessment.
